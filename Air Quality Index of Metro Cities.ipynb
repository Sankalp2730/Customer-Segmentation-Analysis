```python
# ================================================================
# ENVIRONMENTAL ENGINEERING PROJECT
# MUMBAI AIR QUALITY ASSESSMENT AND MITIGATION DECISION SYSTEM
# ================================================================
#
# PURPOSE:
# 1. Analyze air quality of Mumbai
# 2. Identify concerning pollutants
# 3. Identify likely environmental factors
# 4. Suggest multiple mitigation strategies
# 5. Compare strategies using Weighted Sum Model
# 6. Select the best mitigation strategy
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ================================================================
# 1. PROJECT INFORMATION
# ================================================================

CITY = "Mumbai"
STATE = "Maharashtra"
COUNTRY = "India"
YEAR = "FY 2024-25"

print("=" * 85)
print("       ENVIRONMENTAL ENGINEERING")
print(" MUMBAI AIR QUALITY ASSESSMENT & MITIGATION DECISION SYSTEM")
print("=" * 85)

print(f"\nStudy Area : {CITY}, {STATE}, {COUNTRY}")
print(f"Data Period: {YEAR}")

# ================================================================
# 2. MUMBAI AIR QUALITY DATA
# ================================================================
#
# Annual average values used for the project demonstration.
#
# PM2.5 = 39 µg/m³
# PM10  = 83 µg/m³
# O3    = 40 µg/m³
# NO2   = 30 µg/m³
# CO    = 0.7 mg/m³
#
# SO2 is not included in the BMC FY2024-25 summary used here,
# so a representative value is used for demonstration.
# In a real project, replace it with the actual MPCB/CPCB value.
# ================================================================

mumbai_data = {
    "PM2.5": 39,
    "PM10": 83,
    "NO2": 30,
    "SO2": 20,
    "CO": 0.7,
    "O3": 40
}

# Reference limits for project assessment
# Based on Indian ambient-air-quality reference values
pollutant_limits = {
    "PM2.5": 60,
    "PM10": 100,
    "NO2": 80,
    "SO2": 80,
    "CO": 2,
    "O3": 100
}

units = {
    "PM2.5": "µg/m³",
    "PM10": "µg/m³",
    "NO2": "µg/m³",
    "SO2": "µg/m³",
    "CO": "mg/m³",
    "O3": "µg/m³"
}

# ================================================================
# 3. DISPLAY INPUT DATA
# ================================================================

print("\n" + "=" * 85)
print("MUMBAI AIR QUALITY DATA")
print("=" * 85)

input_df = pd.DataFrame({
    "Pollutant": list(mumbai_data.keys()),
    "Concentration": list(mumbai_data.values()),
    "Unit": [units[p] for p in mumbai_data]
})

display(input_df)

# ================================================================
# 4. AIR QUALITY ASSESSMENT
# ================================================================

assessment = []

for pollutant in mumbai_data:

    measured = mumbai_data[pollutant]
    limit = pollutant_limits[pollutant]

    ratio = measured / limit

    exceedance = measured - limit

    percentage = ratio * 100

    if ratio <= 0.50:
        status = "LOW CONCERN"

    elif ratio <= 0.75:
        status = "MODERATE CONCERN"

    elif ratio <= 1.00:
        status = "ACCEPTABLE"

    elif ratio <= 1.50:
        status = "CONCERNING"

    elif ratio <= 2.00:
        status = "HIGH CONCERN"

    else:
        status = "CRITICAL"

    assessment.append({
        "Pollutant": pollutant,
        "Mumbai Value": measured,
        "Reference Limit": limit,
        "Unit": units[pollutant],
        "Pollution Ratio": round(ratio, 2),
        "% of Limit": round(percentage, 1),
        "Exceedance": round(max(0, exceedance), 2),
        "Status": status
    })

assessment_df = pd.DataFrame(assessment)

# ================================================================
# 5. POLLUTANT CONCERN RANKING
# ================================================================

ranking_df = assessment_df.sort_values(
    by="Pollution Ratio",
    ascending=False
).reset_index(drop=True)

ranking_df["Rank"] = range(1, len(ranking_df) + 1)

print("\n" + "=" * 85)
print("MUMBAI POLLUTANT CONCERN ASSESSMENT")
print("=" * 85)

display(
    ranking_df[
        [
            "Rank",
            "Pollutant",
            "Mumbai Value",
            "Reference Limit",
            "Pollution Ratio",
            "% of Limit",
            "Status"
        ]
    ]
)

# ================================================================
# 6. IDENTIFY MOST CONCERNING POLLUTANTS
# ================================================================

concerning_pollutants = ranking_df[
    ranking_df["Pollution Ratio"] > 0.75
]["Pollutant"].tolist()

high_concern_pollutants = ranking_df[
    ranking_df["Pollution Ratio"] > 1
]["Pollutant"].tolist()

print("\n" + "=" * 85)
print("CURRENT MUMBAI AIR-QUALITY CONCERNS")
print("=" * 85)

if concerning_pollutants:

    for pollutant in concerning_pollutants:

        row = ranking_df[
            ranking_df["Pollutant"] == pollutant
        ].iloc[0]

        print(
            f"\n{pollutant}: {row['Status']}"
        )

        print(
            f"Concentration: {row['Mumbai Value']} "
            f"{row['Unit']}"
        )

        print(
            f"Reference limit: {row['Reference Limit']} "
            f"{row['Unit']}"
        )

else:

    print("No major concern identified.")

# ================================================================
# 7. LIKELY ENVIRONMENTAL SOURCES
# ================================================================

pollutant_sources = {

    "PM2.5": [
        "Vehicular emissions",
        "Industrial combustion",
        "Biomass burning",
        "Waste burning",
        "Secondary atmospheric particles"
    ],

    "PM10": [
        "Construction activities",
        "Road dust",
        "Demolition activities",
        "Industrial dust",
        "Resuspension of dust"
    ],

    "NO2": [
        "Vehicular emissions",
        "Diesel vehicles",
        "Traffic congestion",
        "Industrial combustion",
        "Power generation"
    ],

    "SO2": [
        "Industrial fuel combustion",
        "Coal combustion",
        "Oil and fuel burning",
        "Power generation"
    ],

    "CO": [
        "Incomplete combustion",
        "Vehicle emissions",
        "Biomass burning",
        "Fuel combustion"
    ],

    "O3": [
        "Photochemical reactions",
        "NOx emissions",
        "Volatile organic compounds",
        "Strong sunlight",
        "Atmospheric chemical reactions"
    ]
}

print("\n" + "=" * 85)
print("LIKELY POLLUTION SOURCES IN MUMBAI")
print("=" * 85)

source_scores = {}

for pollutant in concerning_pollutants:

    ratio = float(
        ranking_df[
            ranking_df["Pollutant"] == pollutant
        ]["Pollution Ratio"].iloc[0]
    )

    print(f"\n{pollutant}")

    for source in pollutant_sources[pollutant]:

        print("  •", source)

        if source not in source_scores:
            source_scores[source] = 0

        source_scores[source] += ratio

# ================================================================
# 8. ENVIRONMENTAL FACTOR PRIORITY
# ================================================================

source_ranking = pd.DataFrame(
    list(source_scores.items()),
    columns=[
        "Environmental Factor",
        "Concern Score"
    ]
)

if not source_ranking.empty:

    source_ranking = source_ranking.sort_values(
        by="Concern Score",
        ascending=False
    ).reset_index(drop=True)

    source_ranking["Rank"] = range(
        1,
        len(source_ranking) + 1
    )

    print("\n" + "=" * 85)
    print("MUMBAI ENVIRONMENTAL FACTOR PRIORITY")
    print("=" * 85)

    display(
        source_ranking.head(10)
    )

# ================================================================
# 9. MITIGATION STRATEGIES
# ================================================================

strategies = {

    "Electric & Public Transport Expansion": {
        "Effectiveness": 9,
        "Cost": 5,
        "Feasibility": 7,
        "Environmental Impact": 9,
        "Scalability": 9,
        "Sustainability": 9
    },

    "Industrial Emission Control Systems": {
        "Effectiveness": 9,
        "Cost": 6,
        "Feasibility": 8,
        "Environmental Impact": 9,
        "Scalability": 8,
        "Sustainability": 8
    },

    "Construction Dust Management": {
        "Effectiveness": 8,
        "Cost": 8,
        "Feasibility": 9,
        "Environmental Impact": 8,
        "Scalability": 8,
        "Sustainability": 7
    },

    "Open Waste & Biomass Burning Control": {
        "Effectiveness": 8,
        "Cost": 9,
        "Feasibility": 7,
        "Environmental Impact": 9,
        "Scalability": 8,
        "Sustainability": 8
    },

    "Urban Green Infrastructure": {
        "Effectiveness": 6,
        "Cost": 7,
        "Feasibility": 8,
        "Environmental Impact": 10,
        "Scalability": 8,
        "Sustainability": 10
    },

    "Smart Air Quality Monitoring Network": {
        "Effectiveness": 6,
        "Cost": 7,
        "Feasibility": 9,
        "Environmental Impact": 8,
        "Scalability": 10,
        "Sustainability": 9
    }
}

strategy_df = pd.DataFrame(strategies).T

print("\n" + "=" * 85)
print("PROPOSED MITIGATION STRATEGIES")
print("=" * 85)

display(strategy_df)

# ================================================================
# 10. DECISION-MAKING CRITERIA
# ================================================================

weights = {

    "Effectiveness": 0.30,
    "Cost": 0.15,
    "Feasibility": 0.15,
    "Environmental Impact": 0.15,
    "Scalability": 0.10,
    "Sustainability": 0.15
}

print("\n" + "=" * 85)
print("WEIGHTED DECISION-MAKING CRITERIA")
print("=" * 85)

for criterion, weight in weights.items():

    print(
        f"{criterion:30} : "
        f"{weight * 100:.0f}%"
    )

# ================================================================
# 11. CALCULATE WEIGHTED SCORES
# ================================================================

strategy_df["Final Score"] = 0.0

for criterion, weight in weights.items():

    strategy_df["Final Score"] += (
        strategy_df[criterion] * weight
    )

strategy_df["Final Score"] = strategy_df[
    "Final Score"
].round(2)

# ================================================================
# 12. FINAL STRATEGY RANKING
# ================================================================

strategy_ranking = strategy_df.sort_values(
    by="Final Score",
    ascending=False
).copy()

strategy_ranking["Rank"] = range(
    1,
    len(strategy_ranking) + 1
)

print("\n" + "=" * 85)
print("MITIGATION STRATEGY RANKING")
print("=" * 85)

display(
    strategy_ranking[
        [
            "Rank",
            "Effectiveness",
            "Cost",
            "Feasibility",
            "Environmental Impact",
            "Scalability",
            "Sustainability",
            "Final Score"
        ]
    ]
)

# ================================================================
# 13. BEST STRATEGY
# ================================================================

best_strategy = strategy_ranking.index[0]
best_score = strategy_ranking.iloc[0]["Final Score"]

second_strategy = strategy_ranking.index[1]
second_score = strategy_ranking.iloc[1]["Final Score"]

third_strategy = strategy_ranking.index[2]
third_score = strategy_ranking.iloc[2]["Final Score"]

print("\n" + "=" * 85)
print("FINAL MITIGATION DECISION FOR MUMBAI")
print("=" * 85)

print("\n🥇 BEST STRATEGY")
print("-" * 60)

print(best_strategy)
print(f"Weighted Score: {best_score}/10")

print("\n🥈 SECOND BEST STRATEGY")
print("-" * 60)

print(second_strategy)
print(f"Weighted Score: {second_score}/10")

print("\n🥉 THIRD BEST STRATEGY")
print("-" * 60)

print(third_strategy)
print(f"Weighted Score: {third_score}/10")

# ================================================================
# 14. RECOMMENDED COMBINATION
# ================================================================

print("\n" + "=" * 85)
print("RECOMMENDED INTEGRATED STRATEGY")
print("=" * 85)

print(f"\n1. {best_strategy}")
print(f"2. {second_strategy}")
print(f"3. {third_strategy}")

print("""
The recommended approach is an integrated strategy.

Reason:
Mumbai's air pollution is influenced by multiple sources.
Therefore, controlling only one source may not be sufficient.

The top-ranked strategies should be implemented together
to address transportation emissions, industrial emissions,
construction dust and other urban pollution sources.
""")

# ================================================================
# 15. POLLUTANT VISUALIZATION
# ================================================================

plt.figure(figsize=(11, 6))

x = np.arange(len(assessment_df))
width = 0.35

plt.bar(
    x - width/2,
    assessment_df["Mumbai Value"],
    width,
    label="Mumbai Annual Average"
)

plt.bar(
    x + width/2,
    assessment_df["Reference Limit"],
    width,
    label="Reference Limit"
)

plt.xticks(
    x,
    assessment_df["Pollutant"]
)

plt.xlabel("Pollutant")
plt.ylabel("Concentration")

plt.title(
    "Mumbai Air Quality: Pollutant Concentration vs Reference Limit"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()

# ================================================================
# 16. POLLUTION RATIO GRAPH
# ================================================================

plt.figure(figsize=(10, 6))

plt.bar(
    ranking_df["Pollutant"],
    ranking_df["Pollution Ratio"]
)

plt.axhline(
    y=1,
    linestyle="--",
    linewidth=2,
    label="Reference Limit"
)

plt.xlabel("Pollutant")
plt.ylabel("Pollution Ratio")

plt.title(
    "Mumbai Pollutant Concern Level"
)

plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()

# ================================================================
# 17. MITIGATION STRATEGY GRAPH
# ================================================================

plt.figure(figsize=(11, 7))

plot_strategy = strategy_ranking.sort_values(
    by="Final Score"
)

plt.barh(
    plot_strategy.index,
    plot_strategy["Final Score"]
)

plt.xlabel(
    "Weighted Decision Score (0-10)"
)

plt.ylabel(
    "Mitigation Strategy"
)

plt.title(
    "Mumbai Air-Pollution Mitigation Strategy Ranking"
)

plt.xlim(0, 10)

plt.grid(
    axis="x",
    alpha=0.3
)

plt.show()

# ================================================================
# 18. FINAL AUTOMATED REPORT
# ================================================================

print("\n" + "=" * 85)
print("             FINAL MUMBAI ENVIRONMENTAL REPORT")
print("=" * 85)

print(f"\nStudy Area: {CITY}, {STATE}")
print(f"Assessment Period: {YEAR}")

print("\n1. AIR QUALITY ASSESSMENT")
print("-" * 60)

for _, row in ranking_df.iterrows():

    print(
        f"{row['Pollutant']:6} : "
        f"{row['Mumbai Value']} {row['Unit']} "
        f"-> {row['Status']}"
    )

print("\n2. MAJOR CONCERNING POLLUTANTS")
print("-" * 60)

if concerning_pollutants:

    print(
        ", ".join(concerning_pollutants)
    )

else:

    print(
        "No major concern detected."
    )

print("\n3. POLLUTANTS EXCEEDING REFERENCE LIMIT")
print("-" * 60)

if high_concern_pollutants:

    print(
        ", ".join(high_concern_pollutants)
    )

else:

    print(
        "No pollutant exceeds the selected reference limits."
    )

print("\n4. TOP ENVIRONMENTAL FACTORS")
print("-" * 60)

if not source_ranking.empty:

    for _, row in source_ranking.head(5).iterrows():

        print(
            f"{int(row['Rank'])}. "
            f"{row['Environmental Factor']}"
        )

print("\n5. BEST MITIGATION STRATEGY")
print("-" * 60)

print(best_strategy)

print(
    f"Score: {best_score}/10"
)

print("\n6. RECOMMENDED INTEGRATED APPROACH")
print("-" * 60)

print(
    f"{best_strategy} + "
    f"{second_strategy} + "
    f"{third_strategy}"
)

print("\n7. FINAL CONCLUSION")
print("-" * 60)

print("""
The Mumbai Air Quality Assessment and Mitigation Decision
System evaluates important air pollutants and determines
which pollutants require greater attention.

The system then links the concerning pollutants with
potential environmental sources such as transportation,
construction, industrial activities and combustion.

Multiple mitigation strategies are evaluated using a
Weighted Sum Model.

The model considers:

• Effectiveness
• Implementation Cost
• Technical Feasibility
• Environmental Impact
• Scalability
• Long-Term Sustainability

The strategy with the highest weighted score is selected
as the best mitigation strategy.

Because Mumbai's air pollution has multiple contributing
sources, an integrated combination of the highest-ranked
strategies is recommended rather than relying on a single
intervention.

This provides an environmental-engineering decision-support
approach for improving urban air quality in Mumbai.
""")

print("=" * 85)
print("             MUMBAI ANALYSIS COMPLETED")
print("=" * 85)
```

```python
# ================================================================
# ENVIRONMENTAL ENGINEERING PROJECT
# AIR QUALITY ASSESSMENT & MITIGATION DECISION SYSTEM
# COMPARISON OF BENGALURU, DELHI, KOLKATA & MUMBAI
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("=" * 80)
print("   AIR QUALITY ASSESSMENT AND MITIGATION DECISION SYSTEM")
print("   Bengaluru | Delhi | Kolkata | Mumbai")
print("=" * 80)

# ================================================================
# 1. CURRENT CITY DATA
# Date: 02 September 2026
# Values are representative current readings.
# AQI is independently calculated using CPCB breakpoints.
# ================================================================

cities = {
    "Bengaluru": {
        "PM2.5": 30.22,
        "PM10": 52.11,
        "NO2": 16.00,
        "SO2": 8.33,
        "CO": 0.02144,
        "O3": 12.78
    },

    "Delhi": {
        "PM2.5": 59.69,
        "PM10": 53.26,
        "NO2": 7.77,
        "SO2": 5.67,
        "CO": 0.01010,
        "O3": 6.51
    },

    "Kolkata": {
        "PM2.5": 68.00,
        "PM10": 72.43,
        "NO2": 25.14,
        "SO2": 6.00,
        "CO": 0.02629,
        "O3": 18.29
    },

    "Mumbai": {
        "PM2.5": 48.04,
        "PM10": 51.89,
        "NO2": 6.50,
        "SO2": 3.18,
        "CO": 0.00993,
        "O3": 5.18
    }
}

# ================================================================
# 2. CPCB AQI BREAKPOINTS
# AQI ranges:
# 0-50       Good
# 51-100     Satisfactory
# 101-200    Moderately Polluted
# 201-300    Poor
# 301-400    Very Poor
# 401-500    Severe
#
# PM2.5 and PM10 : µg/m³
# NO2, SO2, O3   : µg/m³
# CO              : mg/m³
# ================================================================

breakpoints = {

    "PM2.5": [
        (0, 30, 0, 50),
        (31, 60, 51, 100),
        (61, 90, 101, 200),
        (91, 120, 201, 300),
        (121, 250, 301, 400),
        (251, 500, 401, 500)
    ],

    "PM10": [
        (0, 50, 0, 50),
        (51, 100, 51, 100),
        (101, 250, 101, 200),
        (251, 350, 201, 300),
        (351, 430, 301, 400),
        (431, 600, 401, 500)
    ],

    "NO2": [
        (0, 40, 0, 50),
        (41, 80, 51, 100),
        (81, 180, 101, 200),
        (181, 280, 201, 300),
        (281, 400, 301, 400),
        (401, 800, 401, 500)
    ],

    "SO2": [
        (0, 40, 0, 50),
        (41, 80, 51, 100),
        (81, 380, 101, 200),
        (381, 800, 201, 300),
        (801, 1600, 301, 400),
        (1601, 2000, 401, 500)
    ],

    "CO": [
        (0, 1.0, 0, 50),
        (1.1, 2.0, 51, 100),
        (2.1, 10, 101, 200),
        (10.1, 17, 201, 300),
        (17.1, 34, 301, 400),
        (34.1, 50, 401, 500)
    ],

    "O3": [
        (0, 50, 0, 50),
        (51, 100, 51, 100),
        (101, 168, 101, 200),
        (169, 208, 201, 300),
        (209, 748, 301, 400),
        (749, 1000, 401, 500)
    ]
}

# ================================================================
# 3. AQI CATEGORY
# ================================================================

def aqi_category(aqi):

    if aqi <= 50:
        return "Good"
    elif aqi <= 100:
        return "Satisfactory"
    elif aqi <= 200:
        return "Moderately Polluted"
    elif aqi <= 300:
        return "Poor"
    elif aqi <= 400:
        return "Very Poor"
    else:
        return "Severe"

# ================================================================
# 4. CALCULATE SUB-INDEX
# ================================================================

def calculate_subindex(pollutant, concentration):

    for c_low, c_high, i_low, i_high in breakpoints[pollutant]:

        if c_low <= concentration <= c_high:

            sub_index = (
                ((i_high - i_low) / (c_high - c_low))
                * (concentration - c_low)
                + i_low
            )

            return round(sub_index, 2)

    return 500


# ================================================================
# 5. CALCULATE AQI FOR ALL CITIES
# ================================================================

results = []
subindex_data = {}

for city, values in cities.items():

    city_subindices = {}

    for pollutant, concentration in values.items():

        city_subindices[pollutant] = calculate_subindex(
            pollutant,
            concentration
        )

    overall_aqi = max(city_subindices.values())

    main_pollutant = max(
        city_subindices,
        key=city_subindices.get
    )

    category = aqi_category(overall_aqi)

    subindex_data[city] = city_subindices

    results.append({
        "City": city,
        "AQI": round(overall_aqi, 2),
        "Category": category,
        "Main Pollutant": main_pollutant
    })

aqi_df = pd.DataFrame(results)

# ================================================================
# 6. DISPLAY POLLUTANT CONCENTRATIONS
# ================================================================

print("\n" + "=" * 80)
print("CURRENT POLLUTANT CONCENTRATIONS")
print("=" * 80)

pollutant_df = pd.DataFrame(cities).T

print(pollutant_df.round(2).to_string())

print("\nUnits:")
print("PM2.5, PM10, NO2, SO2, O3 = µg/m³")
print("CO = mg/m³")

# ================================================================
# 7. AQI COMPARISON
# ================================================================

print("\n" + "=" * 80)
print("AQI COMPARISON")
print("=" * 80)

aqi_df = aqi_df.sort_values(
    by="AQI",
    ascending=False
).reset_index(drop=True)

print(aqi_df.to_string(index=False))

# ================================================================
# 8. DETAILED SUB-INDEX RESULTS
# ================================================================

print("\n" + "=" * 80)
print("POLLUTANT-WISE AQI SUB-INDEX")
print("=" * 80)

subindex_df = pd.DataFrame(subindex_data).T

print(subindex_df.round(2).to_string())

# ================================================================
# 9. POLLUTION CONCERN ANALYSIS
# ================================================================

print("\n" + "=" * 80)
print("POLLUTION CONCERN ANALYSIS")
print("=" * 80)

source_mapping = {

    "PM2.5":
        "Vehicle emissions, combustion, industrial emissions, biomass burning",

    "PM10":
        "Construction dust, road dust, demolition and resuspended dust",

    "NO2":
        "Traffic emissions, diesel vehicles and fuel combustion",

    "SO2":
        "Industrial fuel combustion and sulfur-containing fuels",

    "CO":
        "Incomplete combustion and vehicle emissions",

    "O3":
        "Photochemical reactions involving NOx and VOCs"
}

for city in cities:

    print("\n" + "-" * 70)
    print(city)

    city_scores = subindex_data[city]

    ranked = sorted(
        city_scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    print("Pollutants ranked by AQI contribution:")

    for pollutant, score in ranked:

        print(
            f"{pollutant:7s} -> AQI Sub-index: {score:6.2f}"
            f" | Likely source: {source_mapping[pollutant]}"
        )

# ================================================================
# 10. MITIGATION STRATEGIES
# ================================================================

strategies = {

    "Electric & Public Transport Expansion": {
        "Effectiveness": 9,
        "Cost Efficiency": 7,
        "Feasibility": 8,
        "Environmental Impact": 9,
        "Scalability": 9,
        "Sustainability": 10
    },

    "Industrial Emission Control": {
        "Effectiveness": 9,
        "Cost Efficiency": 6,
        "Feasibility": 7,
        "Environmental Impact": 9,
        "Scalability": 8,
        "Sustainability": 9
    },

    "Construction Dust Management": {
        "Effectiveness": 8,
        "Cost Efficiency": 9,
        "Feasibility": 9,
        "Environmental Impact": 8,
        "Scalability": 9,
        "Sustainability": 8
    },

    "Waste & Biomass Burning Control": {
        "Effectiveness": 8,
        "Cost Efficiency": 9,
        "Feasibility": 8,
        "Environmental Impact": 9,
        "Scalability": 8,
        "Sustainability": 9
    },

    "Urban Green Infrastructure": {
        "Effectiveness": 6,
        "Cost Efficiency": 7,
        "Feasibility": 7,
        "Environmental Impact": 10,
        "Scalability": 8,
        "Sustainability": 10
    },

    "Smart Air Quality Monitoring": {
        "Effectiveness": 7,
        "Cost Efficiency": 8,
        "Feasibility": 9,
        "Environmental Impact": 7,
        "Scalability": 10,
        "Sustainability": 9
    }
}

# ================================================================
# 11. WEIGHTED SUM MODEL
# ================================================================

weights = {
    "Effectiveness": 0.30,
    "Cost Efficiency": 0.15,
    "Feasibility": 0.15,
    "Environmental Impact": 0.15,
    "Scalability": 0.10,
    "Sustainability": 0.15
}

strategy_results = []

for strategy, scores in strategies.items():

    score = (
        scores["Effectiveness"] * weights["Effectiveness"] +
        scores["Cost Efficiency"] * weights["Cost Efficiency"] +
        scores["Feasibility"] * weights["Feasibility"] +
        scores["Environmental Impact"] * weights["Environmental Impact"] +
        scores["Scalability"] * weights["Scalability"] +
        scores["Sustainability"] * weights["Sustainability"]
    )

    strategy_results.append({
        "Strategy": strategy,
        "Final Score": round(score, 2)
    })

strategy_df = pd.DataFrame(strategy_results)

strategy_df = strategy_df.sort_values(
    by="Final Score",
    ascending=False
).reset_index(drop=True)

strategy_df["Rank"] = range(1, len(strategy_df) + 1)

# ================================================================
# 12. DISPLAY STRATEGY RANKING
# ================================================================

print("\n" + "=" * 80)
print("MITIGATION STRATEGY RANKING")
print("=" * 80)

print(
    strategy_df[
        ["Rank", "Strategy", "Final Score"]
    ].to_string(index=False)
)

# ================================================================
# 13. BEST STRATEGY
# ================================================================

best_strategy = strategy_df.iloc[0]["Strategy"]
second_strategy = strategy_df.iloc[1]["Strategy"]
third_strategy = strategy_df.iloc[2]["Strategy"]

print("\n" + "=" * 80)
print("RECOMMENDED MITIGATION PLAN")
print("=" * 80)

print("Best Strategy      :", best_strategy)
print("Second Strategy    :", second_strategy)
print("Third Strategy     :", third_strategy)

print("\nRecommended combination:")
print("1.", best_strategy)
print("2.", second_strategy)
print("3.", third_strategy)

# ================================================================
# 14. CITY-SPECIFIC RECOMMENDATIONS
# ================================================================

print("\n" + "=" * 80)
print("CITY-SPECIFIC MITIGATION RECOMMENDATIONS")
print("=" * 80)

for city in cities:

    main_pollutant = aqi_df[
        aqi_df["City"] == city
    ]["Main Pollutant"].values[0]

    print(f"\n{city}:")
    print("Main concern:", main_pollutant)

    if main_pollutant == "PM2.5":
        print("Recommendation: Reduce vehicle emissions, combustion and open burning.")

    elif main_pollutant == "PM10":
        print("Recommendation: Control construction dust, road dust and demolition emissions.")

    elif main_pollutant == "NO2":
        print("Recommendation: Improve public transport and reduce traffic-related emissions.")

    elif main_pollutant == "SO2":
        print("Recommendation: Strengthen industrial emission controls and cleaner fuels.")

    elif main_pollutant == "CO":
        print("Recommendation: Control incomplete combustion and vehicle emissions.")

    elif main_pollutant == "O3":
        print("Recommendation: Control NOx and VOC precursor emissions.")

# ================================================================
# 15. GRAPH 1 - AQI COMPARISON
# ================================================================

plt.figure(figsize=(10, 6))

plt.bar(
    aqi_df["City"],
    aqi_df["AQI"]
)

plt.axhline(
    50,
    linestyle="--",
    label="Good limit (50)"
)

plt.axhline(
    100,
    linestyle="--",
    label="Satisfactory limit (100)"
)

plt.axhline(
    200,
    linestyle="--",
    label="Moderately Polluted limit (200)"
)

plt.title(
    "Air Quality Index Comparison - Bengaluru, Delhi, Kolkata & Mumbai"
)

plt.xlabel("City")
plt.ylabel("AQI")

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()

# ================================================================
# 16. GRAPH 2 - PM2.5 COMPARISON
# ================================================================

plt.figure(figsize=(10, 6))

plt.bar(
    pollutant_df.index,
    pollutant_df["PM2.5"]
)

plt.axhline(
    60,
    linestyle="--",
    label="CPCB PM2.5 AQI breakpoint"
)

plt.title(
    "PM2.5 Concentration Comparison"
)

plt.xlabel("City")
plt.ylabel("PM2.5 (µg/m³)")

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()

# ================================================================
# 17. GRAPH 3 - PM10 COMPARISON
# ================================================================

plt.figure(figsize=(10, 6))

plt.bar(
    pollutant_df.index,
    pollutant_df["PM10"]
)

plt.axhline(
    100,
    linestyle="--",
    label="CPCB PM10 AQI breakpoint"
)

plt.title(
    "PM10 Concentration Comparison"
)

plt.xlabel("City")
plt.ylabel("PM10 (µg/m³)")

plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()

# ================================================================
# 18. GRAPH 4 - MITIGATION STRATEGIES
# ================================================================

plt.figure(figsize=(12, 7))

plt.barh(
    strategy_df["Strategy"],
    strategy_df["Final Score"]
)

plt.xlabel("Weighted Decision Score")
plt.ylabel("Mitigation Strategy")

plt.title(
    "Mitigation Strategy Ranking"
)

plt.grid(axis="x", alpha=0.3)

plt.gca().invert_yaxis()

plt.show()

# ================================================================
# 19. FINAL REPORT
# ================================================================

print("\n" + "=" * 80)
print("FINAL ENVIRONMENTAL ENGINEERING REPORT")
print("=" * 80)

highest_city = aqi_df.iloc[0]["City"]
lowest_city = aqi_df.iloc[-1]["City"]

print("\n1. AIR QUALITY COMPARISON")
print("-------------------------")

print(
    f"Highest AQI: {highest_city} "
    f"({aqi_df.iloc[0]['AQI']})"
)

print(
    f"Lowest AQI: {lowest_city} "
    f"({aqi_df.iloc[-1]['AQI']})"
)

print("\n2. CITY RANKING")
print("----------------")

for i, row in aqi_df.iterrows():

    print(
        f"{i+1}. {row['City']} "
        f"-> AQI {row['AQI']} "
        f"-> {row['Category']} "
        f"-> Main pollutant: {row['Main Pollutant']}"
    )

print("\n3. MAJOR POLLUTION FACTORS")
print("----------------------------")

print("• Vehicular emissions")
print("• Construction and road dust")
print("• Industrial emissions")
print("• Waste and biomass burning")
print("• Fuel combustion")
print("• Urbanization")
print("• Meteorological conditions")

print("\n4. BEST MITIGATION STRATEGY")
print("----------------------------")

print(best_strategy)

print("\n5. RECOMMENDED COMBINATION")
print("---------------------------")

print("•", best_strategy)
print("•", second_strategy)
print("•", third_strategy)

print("\n6. CONCLUSION")
print("--------------")

print(
    f"The comparison shows that {highest_city} has the highest "
    f"current AQI among the four cities in this dataset, while "
    f"{lowest_city} has the lowest."
)

print(
    "The decision-making framework evaluates mitigation options "
    "using effectiveness, cost efficiency, technical feasibility, "
    "environmental impact, scalability and long-term sustainability."
)

print(
    "A combined mitigation approach is recommended instead of relying "
    "on a single intervention because urban air pollution has multiple sources."
)

print("\n" + "=" * 80)
print("PROJECT EXECUTION COMPLETED")
print("=" * 80)
```
